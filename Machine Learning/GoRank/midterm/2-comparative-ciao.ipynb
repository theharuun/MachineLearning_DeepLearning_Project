# ==========================================
# PROGRESS 1: DATA LOADING & PREPROCESSING (.mat)
# ==========================================

```python
import pandas as pd
import numpy as np
import scipy.io as sio
from sklearn.model_selection import train_test_split


# 1. Load Ciao rating.mat file using scipy.io
file_path_ciao = r"C:\Users\Harun\Desktop\üni\4.SINIF\2.donem\recommender\dataset\ciao\rating.mat"
mat_data_ciao = sio.loadmat(file_path_ciao)

# .mat file returns a dictionary. Actual data is usually stored under 'rating' key.
# Convert the (N, 5) numpy array into a Pandas DataFrame
columns_tang = ['user_id', 'item_id', 'category_id', 'rating', 'helpfulness']
df_ciao = pd.DataFrame(mat_data_ciao['rating'], columns=columns_tang)

# IDs may come as float from MATLAB, convert to integer (int)
df_ciao['user_id'] = df_ciao['user_id'].astype(int)
df_ciao['item_id'] = df_ciao['item_id'].astype(int)

# Only User, Item and Rating are needed. Drop others.
df_ciao = df_ciao[['user_id', 'item_id', 'rating']]

# Calculate statistics
n_users_ciao = df_ciao['user_id'].nunique()
n_items_ciao = df_ciao['item_id'].nunique()
n_ratings_ciao = len(df_ciao)
sparsity_ciao = 1.0 - (n_ratings_ciao / (n_users_ciao * n_items_ciao))

print("--- CIAO - PROGRESS 1 ---")
print(f"Total Users: {n_users_ciao}")
print(f"Total Products: {n_items_ciao}")
print(f"Total Ratings: {n_ratings_ciao}")
print(f"Sparsity Ratio: {sparsity_ciao:.6f} ({sparsity_ciao*100:.4f}%)")
print("\nFirst 5 Rows of Simplified Dataset:")
display(df_ciao.head())
```

# ==========================================
# PROGRESS 2: TRAIN/TEST SPLIT & MATRIX
# ==========================================

```python
from sklearn.model_selection import train_test_split
import numpy as np

# 1. We split the data into 80% Train and 20% Test
train_ciao, test_ciao = train_test_split(df_ciao, test_size=0.20, random_state=42)

print("\n--- CIAO - PROGRESS 2 ---")
print(f"Training Set Size: {len(train_ciao)} rows")
print(f"Test Set Size: {len(test_ciao)} rows\n")

# BUG SOLUTION: We use pivot_table instead of pivot for duplicate votes.
# We take the average of the votes that come to the same product/user intersection with aggfunc='mean'. 
user_item_matrix_ciao = train_ciao.pivot_table(
    index='user_id', 
    columns='item_id', 
    values='rating',
    aggfunc='mean' 
)

print(f"User-Product Matrix Size: {user_item_matrix_ciao.shape}")
print("Appearance of the Generated Matrix (Spaces NaN):")
display(user_item_matrix_ciao.head())
```

```python
from surprise import Reader, Dataset, SVD
from surprise.model_selection import train_test_split as surprise_train_test_split
from surprise import accuracy

print("--- CIAO: SVD MODEL TRAINING AND EVALUATION ---")

# 1. Prepare Data for Surprise
# Ciao ratings are 1 to 5
reader = Reader(rating_scale=(1, 5))
data_ciao = Dataset.load_from_df(user_item_matrix_ciao.reset_index().melt(id_vars='user_id', value_name='rating').dropna(), reader)

# 2. Train/Test Split
trainset_ciao, testset_ciao = surprise_train_test_split(data_ciao, test_size=0.20, random_state=42)

# 3. Train SVD
svd_model_ciao = SVD(n_factors=100, random_state=42)
svd_model_ciao.fit(trainset_ciao)

# 4. Predict and Evaluate
predictions_ciao = svd_model_ciao.test(testset_ciao)

print("\nResults for Ciao Dataset:")
rmse_ciao = accuracy.rmse(predictions_ciao, verbose=True)
mae_ciao = accuracy.mae(predictions_ciao, verbose=True)
```
